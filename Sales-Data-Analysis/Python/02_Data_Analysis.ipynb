EDA — Exploratory Data Analysis

```python
import pandas as pd
import numpy as np
df=pd.read_csv("Cleaned_SampleSuperstore.csv")
print(df)
```

```python
total_sales=df["Sales"].sum()
total_quantity=df["Quantity"].sum()
total_profit=df["Profit"].sum()
print("Total Sales=",total_sales)
print("Total Quantity=",total_quantity)
print("Total Profit=",total_profit)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
print(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].sum()
print(category_profit)
```

```python
category_analysis = df.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})

category_analysis["Profit Margin %"] = (category_analysis["Profit"] / category_analysis["Sales"]) * 100

print(category_analysis)
```

```python
furniture_analysis = df[df["Category"] == "Furniture"].groupby("Sub-Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})

furniture_analysis["Profit Margin %"] = (
    furniture_analysis["Profit"] / furniture_analysis["Sales"]
) * 100

print(furniture_analysis.sort_values("Profit", ascending=True))
```

```python
tables_analysis = df[df["Sub-Category"] == "Tables"].groupby("Discount").agg({
    "Sales": "sum",
    "Profit": "sum"
})

tables_analysis["Profit Margin %"] = (
    tables_analysis["Profit"] / tables_analysis["Sales"]
) * 100

print(tables_analysis)
```

```python
region_analysis = df.groupby("Region").agg({
    "Sales": "sum",
    "Profit": "sum"
})

region_analysis["Profit Margin %"] = (
    region_analysis["Profit"] / region_analysis["Sales"]
) * 100

print(region_analysis.sort_values("Sales", ascending=False))
```

```python
segment_analysis = df.groupby("Segment").agg({
    "Sales": "sum",
    "Profit": "sum"
})

segment_analysis["Profit Margin %"] = (
    segment_analysis["Profit"] / segment_analysis["Sales"]
) * 100

print(segment_analysis.sort_values("Sales", ascending=False))
```

```python
subcategory_analysis = df.groupby("Sub-Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})

subcategory_analysis["Profit Margin %"] = (
    subcategory_analysis["Profit"] / subcategory_analysis["Sales"]
) * 100

print(subcategory_analysis.sort_values("Profit", ascending=False))
```

```python
city_sales = df.groupby("City")["Sales"].sum()
print(city_sales.sort_values(ascending=False).head(10))
```

```python
city_profit = df.groupby("City")["Profit"].sum()
print(city_profit.sort_values(ascending=False).head(10))
```

```python
discount_analysis = df.groupby("Discount").agg({
    "Sales": "sum",
    "Profit": "sum"
})

discount_analysis["Profit Margin %"] = (
    discount_analysis["Profit"] / discount_analysis["Sales"]
) * 100

print(discount_analysis)
```

```python
profitable = (df["Profit"] > 0).sum()
loss_making = (df["Profit"] < 0).sum()
zero_profit = (df["Profit"] == 0).sum()

print("Profitable Transactions:", profitable)
print("Loss-making Transactions:", loss_making)
print("Zero-profit Transactions:", zero_profit)
```

```python
#overall average discount
average_discount = df["Discount"].mean()
print("Average Discount =", round(average_discount * 100, 2), "%")
```

```python
#Average Order/Transaction Value
average_sales = df["Sales"].mean()
print("Average Sales per Transaction =", round(average_sales, 2))
```

```python
#Calculate Overall Profit Margin
profit_margin = (total_profit / total_sales) * 100
print("Overall Profit Margin =", round(profit_margin, 2), "%")
```
