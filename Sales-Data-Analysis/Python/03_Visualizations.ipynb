```python
import pandas as pd
import numpy as np
df=pd.read_csv("Cleaned_SampleSuperstore.csv")
print(df)
```

```python
import matplotlib.pyplot as plt
```

```python
#Sales by category
category_sales = df.groupby("Category")["Sales"].sum()
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit by category
category_profit = df.groupby("Category")["Profit"].sum()
category_profit.plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
#visualize Sales vs Profit
category_summary = df.groupby("Category")[["Sales", "Profit"]].sum()
category_summary.plot(kind="bar")
plt.title("Sales vs Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit Margin by Category
category_margin = (
    df.groupby("Category")["Profit"].sum() / df.groupby("Category")["Sales"].sum()) * 100
category_margin.plot(kind="bar")

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Sales by Region
region_sales=df.groupby("Region")["Sales"].sum()
region_sales.plot(kind="bar")

plt.title("Sales by region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit by regions
region_profit=df.groupby("Region")["Profit"].sum()
region_profit.plot(kind="bar")

plt.title("Profit by region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit Margin by Region--->(profit/sales)*100
region_margin=(df.groupby("Region")["Profit"].sum()/df.groupby("Region")["Sales"].sum())*100
region_margin.plot(kind="bar")

plt.title("Profit margin by region")
plt.xlabel("Region")
plt.ylabel("Profit Margin(%)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Sales by Segment
segment_sales=df.groupby("Segment")["Sales"].sum()
segment_sales.plot(kind="bar")

plt.title("Sales by Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit by Segment
segment_profit=df.groupby("Segment")["Profit"].sum()
segment_profit.plot(kind="bar")

plt.title("Profit by Segment")
plt.xlabel("Segment")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
#Profit Margin by Segment
segment_margin=(df.groupby("Segment")["Profit"].sum()/df.groupby("Segment")["Sales"].sum())*100
segment_margin.plot(kind="bar")

plt.title("Profit Margin by Segment")
plt.xlabel("Segment")
plt.ylabel("Profit Margin(%)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Sales by Sub-Category
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
subcategory_sales.plot(kind="bar")

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#Profit by Sub-Category
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)
subcategory_profit.plot(kind="bar")

plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
#Profit Margin by sub-category
subcategory_margin = (df.groupby("Sub-Category")["Profit"].sum()/df.groupby("Sub-Category")["Sales"].sum())*100
subcategory_margin = subcategory_margin.sort_values(ascending=False)
subcategory_margin.plot(kind="bar")

plt.title("Profit Margin by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
#Sales vs Profit by Sub-Category
subcategory_summary = df.groupby("Sub-Category")[["Sales", "Profit"]].sum()
subcategory_summary = subcategory_summary.sort_values(by="Sales", ascending=False)
subcategory_summary.plot(kind="bar", figsize=(12, 6))

plt.title("Sales vs Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.show()
```

Discount Analysis

```python
#Discount vs Profit Margin
# Profit Margin by Discount
discount_margin = (df.groupby("Discount")["Profit"].sum() /df.groupby("Discount")["Sales"].sum()) * 100
discount_margin = discount_margin.sort_index()
discount_margin.plot(kind="bar")

plt.title("Profit Margin by Discount")
plt.xlabel("Discount")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Sales vs Profit by Discount
discount_analysis = df.groupby("Discount")[["Sales", "Profit"]].sum()
discount_analysis.plot(kind="bar")

plt.title("Sales vs Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
# Discount vs Profit
# Profit by Discount
discount_profit = df.groupby("Discount")["Profit"].sum()
discount_profit.plot(kind="bar")

plt.title("Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

Transaction Analysis

```python
#Profitable vs Loss-making Transactions
# Transaction Profitability
profitable = (df["Profit"] > 0).sum()
loss_making = (df["Profit"] < 0).sum()
zero_profit = (df["Profit"] == 0).sum()

transaction_data = {
    "Profitable": profitable,
    "Loss-making": loss_making,
    "Zero Profit": zero_profit
}

transaction_series = pd.Series(transaction_data)
transaction_series.plot(kind="bar")

plt.title("Transaction Profitability")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.show()
```

```python
# Top 10 Cities by Sales

city_sales = df.groupby("City")["Sales"].sum()
city_sales = city_sales.sort_values(ascending=False).head(10)
city_sales.plot(kind="bar",figsize=(10,5))

plt.title("Top 10 Cities by Sales")
plt.xlabel("City")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Top 10 Cities by Profit

city_profit = df.groupby("City")["Profit"].sum()
city_profit = city_profit.sort_values(ascending=False).head(10)
city_profit.plot(kind="bar")

plt.title("Top 10 Cities by Profit")
plt.xlabel("City")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# Worst 10 Sub-Categories by Profit
worst_subcategories = df.groupby("Sub-Category")["Profit"].sum()
worst_subcategories = worst_subcategories.sort_values(ascending=True).head(10)
worst_subcategories.plot(kind="bar")

plt.title("Worst 10 Sub-Categories by Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# Top 10 Sub-Categories by Profit
top_subcategories = df.groupby("Sub-Category")["Profit"].sum()
top_subcategories = top_subcategories.sort_values(ascending=False).head(10)
top_subcategories.plot(kind="bar")

plt.title("Top 10 Sub-Categories by Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```
